```python
lst=[1,2,3,4,5,6,7]
evensum=0
oddsum=0
for i in lst:
  if(i%2==0):
    evensum=evensum+i
  else:
    oddsum=oddsum+i
print("odd sum is{}".format(oddsum))
print(f"Even sum is{evensum}")

```

```python

i=0
evensum=0
oddsum=0
while(i<=10):
  if(i%2==0):
    evensum=evensum+i
  else:
    oddsum=oddsum+i
  i=i+1
print(evensum,oddsum)
```
