```python
data=('index,a,b,c\n'
        '4,apple,bat,5.7\n'
           '8,orange,cow,10')
```

```python
pd.read_csv(StringIO(data),index_col=0)
```

```python
##useindexcol and usecols
pd.read_csv(StringIO(data),usecols=['a','b','c'],index_col=0)
```
